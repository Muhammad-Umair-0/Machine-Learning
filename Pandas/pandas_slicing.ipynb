```python
import pandas as pd
import numpy as np
```

```python
player_list = [['M.S.Dhoni', 36, 75, 5428000],
               ['A.B.D Villers', 38, 74, 3428000],
               ['V.Kohli', 31, 70, 8428000],
               ['S.Smith', 34, 80, 4428000],
               ['C.Gayle', 40, 100, 4528000],
               ['J.Root', 33, 72, 7028000],
               ['K.Peterson', 42, 85, 2528000]]

df = pd.DataFrame(player_list, columns=['Name', 'Age', 'Weight', 'Salary'])
print(df)
```

```python
# slicing using iloc 
df1 = df.iloc[0:4]
df1
```

```python
# slicing column in df
df1 = df.iloc[:, 0:2]
df1
```

```python
# selecting the specific comlum
# iloc[Rows number , columns number]
value = df.iloc[2,3]
print("The specific cell Value is ",value)
```

### Using the Boolean conditions in pandas DataFrame
 Instead of selecting rows by index we can use Boolean conditions (e.g., Age > 35) to filter rows dynamically.
 

```python
data = df[df['Age']>35].iloc[:,:]
print("Fiiltered Data based on age >35 :\n",data)
```

### slicing using loc
Slicing Rows in Dataframe
With loc[] we can extract a range of rows by their labels instead of integer positions

```python
df.set_index('Name', inplace=True)
print("DataFrame with custom index:\n",df)
```

```python
custom = df.loc['A.B.D Villers':'S.Smith']
print(custom)
```

```python
# selecting specific cell 
# df.loc[label, column]
values = df.loc['V.Kohli', 'Salary']
print(values)
```

# Filter Pandas Dataframe with multiple conditions

```python
# assign data
df = pd.DataFrame({'Name': [' RACHEL  ', ' MONICA  ', ' PHOEBE  ',
                                   '  ROSS    ', 'CHANDLER', ' JOEY    '],
                          
                          'Age': [30, 35, 37, 33, 34, 30],
                          
                          'Salary': [100000, 93000, 88000, 120000, 94000, 95000],
                          
                          'JOB': ['DESIGNER', 'CHEF', 'MASUS', 'PALENTOLOGY',
                                  'IT', 'ARTIST']})
```

```python
#  filter dataframe
display(df.loc[(df['Salary']>=100000) & (df['Age']< 40) & (df['JOB'].str.startswith('D')),
                    ['Name','JOB']])
```

```python
# filter with numpy 
Values = np.where((df['Salary']>=100000) & (df['Age']< 50) & (df['JOB'].str.startswith('D')) )
print(Values)
display(df.loc[Values])
```

```python
# Filter Pandas Dataframe Using Query (eval and query works only with columns)


display(df.query('Salary  <= 100000 & Age < 40 & JOB.str.startswith("C").values'))
```
